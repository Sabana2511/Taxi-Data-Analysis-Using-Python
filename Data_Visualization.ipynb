```python
# Taxi Data Analysis Using Python
# Import required libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Set the style for plots
sns.set_style("whitegrid")

# Load the taxis dataset
df = sns.load_dataset("taxis")

# Display first 5 rows
df.head()
```

```python
# Display dataset information
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nFirst 5 Rows:")
display(df.head())

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Check the number of missing values in each column
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)

# Display only columns that contain missing values
print("\nColumns with Missing Values:")
print(missing_values[missing_values > 0])
```

```python
# Visualize missing values
plt.figure(figsize=(12, 5))

sns.heatmap(df.isnull(),
            cbar=False,
            cmap="viridis")

plt.title("Missing Values in Taxi Dataset")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```

```python
# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(include=np.number).columns
categorical_columns = df.select_dtypes(include="object").columns

print("Numerical Columns:")
print(list(numerical_columns))

print("\nCategorical Columns:")
print(list(categorical_columns))
```

```python
# Fill missing numerical values with the median
for col in numerical_columns:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Fill missing categorical values with the mode
for col in categorical_columns:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Check missing values after imputation
print("Missing values after imputation:")
print(df.isnull().sum())
```

```python
# Remove rows with missing values in critical columns
critical_columns = ["pickup", "fare", "distance", "payment", "pickup_borough"]

df = df.dropna(subset=critical_columns)

print("Dataset shape after cleaning:", df.shape)

print("\nRemaining missing values:")
print(df.isnull().sum())
```

```python
# Convert pickup column to datetime
df["pickup"] = pd.to_datetime(df["pickup"])

# Check the data type
print(df["pickup"].dtype)

# Display pickup values
df[["pickup"]].head()
```

```python
#LINE CHART
# Sort data by pickup time
df_sorted = df.sort_values("pickup")

# Create line chart
plt.figure(figsize=(14, 6))

plt.plot(
    df_sorted["pickup"],
    df_sorted["fare"],
    color="blue",
    linewidth=1
)

plt.title("Fare Over Time")
plt.xlabel("Pickup Time")
plt.ylabel("Fare")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
#BAR CHART

# Calculate total fare for each pickup borough
borough_fare = df.groupby("pickup_borough")["fare"].sum().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(10, 6))

borough_fare.plot(
    kind="bar",
    color="steelblue"
)

plt.title("Total Fare by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#PIE CHART
# Count trips by payment method
payment_counts = df["payment"].value_counts()

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=sns.color_palette("Set2")
)

plt.title("Distribution of Trips by Payment Method")
plt.axis("equal")

plt.show()
```

```python
#HISTOGRAM

# Create histogram
plt.figure(figsize=(10, 6))

plt.hist(
    df["distance"],
    bins=30,
    color="skyblue",
    edgecolor="black"
)

plt.title("Distribution of Taxi Trip Distance")
plt.xlabel("Distance")
plt.ylabel("Number of Trips")

plt.tight_layout()
plt.show()
```

```python
  #BOX PLOT
  # Create box plot
plt.figure(figsize=(12, 6))

df.boxplot(
    column="tip",
    by="pickup_borough",
    grid=False
)

plt.title("Distribution of Tips by Pickup Borough")
plt.suptitle("")  # Remove automatic pandas title
plt.xlabel("Pickup Borough")
plt.ylabel("Tip Amount")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#SEABORN VISUALIZATION
#COUNT PLOT
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    x="pickup_borough",
    hue="pickup_borough",
    palette="Set2",
    legend=False
)

plt.title("Number of Taxi Trips by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Number of Trips")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
#SCATTER PLOT
plt.figure(figsize=(12, 7))

sns.scatterplot(
    data=df,
    x="distance",
    y="fare",
    hue="pickup_borough",
    palette="Set1",
    alpha=0.6
)

plt.title("Relationship Between Distance and Fare")
plt.xlabel("Distance")
plt.ylabel("Fare")
plt.legend(title="Pickup Borough")

plt.tight_layout()
plt.show()
```

```python
# Heatmap
# Select numerical columns
correlation_columns = [
    "distance",
    "fare",
    "tip",
    "tolls",
    "total"
]

# Calculate correlation matrix
correlation_matrix = df[correlation_columns].corr()

# Create heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Taxi Numerical Variables")

plt.tight_layout()
plt.show()
```

```python
# Pair Plot
# Select required columns

pair_columns = [
    "distance",
    "fare",
    "tip",
    "total",
    "pickup_zone"
]

pair_df = df[pair_columns].dropna()

# Take a sample for better visualization
pair_sample = pair_df.sample(
    n=min(1000, len(pair_df)),
    random_state=42
)

pair_sample.head()
```

```python
sns.pairplot(
    pair_sample,
    vars=["distance", "fare", "tip", "total"],
    hue="pickup_zone"
)

plt.show()
```

```python
# violin plot

plt.figure(figsize=(10, 6))

sns.violinplot(
    data=df,
    x="payment",
    y="fare"
)

plt.xlabel("Payment Method")
plt.ylabel("Fare")
plt.title("Distribution of Fare by Payment Method")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```
